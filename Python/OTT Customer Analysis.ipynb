```python
platform_counts = df["Platform"].value_counts()

plt.figure(figsize=(10, 6))

plt.bar(
    platform_counts.index,
    platform_counts.values
)

plt.title("Customer Distribution by OTT Platform")
plt.xlabel("OTT Platform")
plt.ylabel("Number of Customers")

plt.xticks(rotation=30)

plt.tight_layout()

plt.show()
```

```python
platform_counts = df["Platform"].value_counts()

plt.figure(figsize=(10, 6))

plt.bar(
    platform_counts.index,
    platform_counts.values
)

plt.title("Customer Distribution by OTT Platform")
plt.xlabel("OTT Platform")
plt.ylabel("Number of Customers")

plt.xticks(rotation=30)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Age",
    bins=15,
    kde=True
)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.show()
```

```python
gender_counts = df["Gender"].value_counts()

print(gender_counts)
```

```python
gender_percentage = (
    df["Gender"]
    .value_counts(normalize=True)
    * 100
)

print(gender_percentage.round(2))
```

```python
subscription_counts = df["SubscriptionType"].value_counts()

print(subscription_counts)
```

```python
plt.figure(figsize=(9, 6))

sns.countplot(
    data=df,
    x="SubscriptionType",
    order=df["SubscriptionType"].value_counts().index
)

plt.title("Subscription Type Distribution")
plt.xlabel("Subscription Type")
plt.ylabel("Number of Customers")

plt.show()
```

```python
location_counts = df["Location"].value_counts()

print(location_counts)
```

```python
plt.figure(figsize=(10, 7))

sns.countplot(
    data=df,
    y="Location",
    order=df["Location"].value_counts().index
)

plt.title("Customer Distribution by Location")
plt.xlabel("Number of Customers")
plt.ylabel("Location")

plt.show()
```
